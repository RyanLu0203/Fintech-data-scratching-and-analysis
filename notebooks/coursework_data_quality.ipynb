This notebook profiles the synthetic Smoke fixture only. It does not characterize real historical market/news coverage. To inspect real inputs, read the four files in docs/data.md with symbol dtype=str, then call the same validators.


```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / "src").is_dir() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
from scripts.run_coursework import synthetic_inputs
from src.evaluation.coursework import validate_market, input_profile
from src.nlp.chronological import normalize_news, map_news_to_sessions
market, target, peers, peer_market = synthetic_inputs()
profile = input_profile(validate_market(market), normalize_news(target), normalize_news(peers))
profile
```

```python
mapped = map_news_to_sessions(normalize_news(target), market["date"])
mapped[["news_id", "symbol", "available_at", "decision_date", "source"]].head(8)
```

```python
daily_coverage = mapped.groupby("decision_date").size().reindex(market["date"], fill_value=0)
daily_coverage.describe()
```

The canonical key is stock × trading session for market data and stock × news identity for text. Duplicate market keys fail validation. Dates with zero stories remain market sessions; a zero news count is missing coverage, not neutral text. Inspect source, available time and class balance before any real-data conclusion.
